```python
### Now we will make predictions with the model we are TRAINING 
model_Y_train = mlr_model.predict(train_data[features])
model_Y_train

#save the predictions as a column in the traning dataset 
train_data['mlr_model_train']= model_Y_train


### Now we will make predictions with the model we are TESTING
model_Y_test = mlr_model.predict(test_data[features])
model_Y_test

#save the predictions as a column in the testing dataset 
test_data['mlr_model_test']= model_Y_test
```

### 3. Lets evaluate the basic model predictions to see how they perform using RMSE. 

```python
### Lets look at the RMSE for the basic model for the TRAINING errors
train_rmse_basic = mean_squared_error(train_data['Energy_Use_kWh'], train_data['mlr_model_train'], squared=False)

print('Training RMSE: {:.4g}'.format(train_rmse_basic))


### Lets look at the RMSE for the basic model for the TESTING errors
test_rmse_basic = mean_squared_error(test_data['Energy_Use_kWh'], test_data['mlr_model_test'], squared=False)

print('Training RMSE: {:.4g}'.format(test_rmse_basic))
```
