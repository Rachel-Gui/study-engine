```python
# lets create a new data frame with the amount of visitors per animal at the zoo

import pandas as pd

zoo = pd.DataFrame({'Animal': ['elephant', 'whale', 'turtle', 'panda'],
                   'Visitors': [150, 200, 100, 300],
                   'Age': [1, 10, 80, 100],
                   'Type': ['land', 'sea', 'both', 'land']})
zoo
```

```python
# lets print the column names with .columns

zoo.columns
```

```python
#lets print or index on one column

zoo.Animal

zoo['Type']
```

```python
#now lets only look at one animal

elephant= zoo[zoo.Animal== 'elephant']

elephant
```

```python
#lets make a dataframe for those who are older

elephant= zoo[zoo.Age>= 50]

elephant
```

```python
#get all of the land animals


land= zoo[zoo['Type']=='land']

land
```

# **11.3. Data Visualization with Matplotlib**

```python
import matplotlib.pyplot as plt

# Simple line plot
x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x, y)
plt.title('Simple Line Plot')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
#lets make a plot of the relationship

plt.scatter(zoo['Age'], zoo['Visitors'])
plt.title('Zoo animal age and visitors');
plt.xlabel('Age')
plt.ylabel('Visitors');
plt.show()

```

# **11.4. Scikit-learn for Machine Learning**

```python
from sklearn.linear_model import LinearRegression

# Example data
X = [[1], [2], [3], [4]]
y = [1, 2, 3, 4]

# Create a LinearRegression model
model = LinearRegression()
model.fit(X, y)

# Make a prediction
prediction = model.predict([[5]])
print(f"Prediction for 5: {prediction[0]}")
```
